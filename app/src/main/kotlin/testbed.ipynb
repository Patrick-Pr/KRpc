```kotlin
import kotlin.reflect.KClass
import kotlin.reflect.KType
import kotlin.reflect.typeOf

val test: KType = typeOf<List<String>>()

val klass = test.classifier as? KClass<*>
println(klass)
println(klass?.simpleName)
println(test.arguments)
println(test.toString())

val str = if (test.arguments.isEmpty()) """
fun doSomething(arg: ${klass?.simpleName}) {
    println(arg)
}
""".trimMargin()
else """
fun doSomething(arg: ${klass?.simpleName}<${
    test.arguments.map { (it.type?.classifier as? KClass<*>)?.simpleName }.joinToString(separator = ", ")
}>) {
    println(arg)
}
""".trimMargin()

println(str)
```
